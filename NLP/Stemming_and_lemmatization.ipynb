#### Stemming and lemmatization 

###### nltk supports both stemming and lemmatization but spacy supports only lemmatization

```python
import nltk
import spacy
```

```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
```

```python
words =[
    "walking", "jumping", "running", "hopping", 
    "walked", "jumped", "planned", "dropped", 
    "walks", "jumps", "ponies", "flies", "stories",   "quickly", "smoothly" ,"happily", 
    "argument", "adjustment", "kindness", "highest" ,"ate"
]
```

```python
for word in words:
    print(word ,"|", stemmer.stem(word))
```

```python
nlp = spacy.load("en_core_web_sm")
doc =nlp( "walking jumping running hopping walked jumped planned dropped walks jumps ponies flies stories quickly smoothly happily argument adjustment kindness highest ate" )
```

```python
for token in doc:
    print(token ,"|" , token.lemma_)
```

```python
# customization og rules for base word 
doc =nlp(" bro brah talking alks talked bruh")
ar = nlp.get_pipe("attribute_ruler")
ar.add([ [{"TEXT":"bro"}],[{"TEXT":"brah"} ],[{"TEXT": "bruh"}]],{"LEMMA": "Brother"})
```

```python
for token in doc:
    print(token ,"|", token.lemma_)
```
