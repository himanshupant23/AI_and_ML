```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
doc = nlp("The quick brown fox jumped over the lazy dog.")
```

```python
for token in doc:
    print(token,"|", token.pos_, "|" ,spacy.explain(token.pos_) ,"|" , token.tag_ ,"|",spacy.explain(token.tag_))
```

###### we can also filters the token means text cleaning by removing punctuation spaces etc

```python
doc = nlp("Wow!  Did Apple Inc. really buy AI_Tech (for $1.2B) at 5:00 PM?!  The data-driven team moved quickly...")
```

```python
for token in doc:
    print(token ,"|", token.pos_ ,"|" , spacy.explain(token.pos_))
```

```python
filtered_token =[]
for token in doc:
    if token.pos_ not in ["SPACE","X","PUNCT"]:
        filtered_token.append(token)
```
