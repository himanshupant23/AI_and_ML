#### Information Extraction 


```python
import re
import warnings 
warnings.filterwarnings('ignore')
```

```python
chat1 = 'codebasics: you ask lot of questions 😠  1235678912, abc@xyz.com'
chat2 = 'codebasics: here it is: (123)-567-8912, abc@xyz.com'
chat3 = 'codebasics: yes, phone: 1235678912 email: abc@xyz.com'
```

```python
 # extracting number from chat 
pattern = "\d{10} | \(\d{3}\)-\d{3}-\d{4}"
matches = re.findall(pattern,chat2)
matches
```

```python
pattern2 = "\(\d{3}\)-\d{3}-\d{4}"
matches2 = re.findall(pattern2,chat2)
matches2
```

```python
# extracting email 
pattern_match = '[a-zA-Z0-9_]*@[a-z]*\.[a-zA-Z0-9]*'
email_match = re.findall(pattern_match,chat2)
email_match
```

```python
chat4 = "codebasics: My order 412889912 is having an issue, I was charged 300$ when online it says 280$"
pattern_match2 = 'order[^\d]*(\d*)'
order_match = re.findall(pattern_match2,chat4)
order_match
```

```python
text='''
Born	Elon Reeve Musk
June 28, 1971 (age 50)
Pretoria, Transvaal, South Africa
Citizenship	
South Africa (1971–present)
Canada (1971–present)
United States (2002–present)
Education	University of Pennsylvania (BS, BA)
Title	
Founder, CEO and Chief Engineer of SpaceX
CEO and product architect of Tesla, Inc.
Founder of The Boring Company and X.com (now part of PayPal)
Co-founder of Neuralink, OpenAI, and Zip2
Spouse(s)	
Justine Wilson
​
​(m. 2000; div. 2008)​
Talulah Riley
​
​(m. 2010; div. 2012)​
​
​(m. 2013; div. 2016)
'''
```

```python
def get_pattern_match(pattern, text):
    matches = re.findall(pattern, text)
    if matches:
        return matches[0]
```

```python
def extract_personal_information(text):
    age = get_pattern_match('age (\d+)', text)
    full_name = get_pattern_match('Born(.*)\n', text)
    birth_date = get_pattern_match('Born.*\n(.*)\(age', text)
    birth_place = get_pattern_match('\(age.*\n(.*)', text)
    return {
        'age': int(age),
        'name': full_name.strip(),
        'birth_date': birth_date.strip(),
        'birth_place': birth_place.strip()
    }
```

```python
extract_personal_information(text)
```
