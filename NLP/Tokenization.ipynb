##### customizing the tokenizers

```python
doc = nlp("gimme double cheese extra large healthy pizza" )
```

```python
from spacy.symbols import ORTH
nlp.tokenizer.add_special_case("gimme", [
                 {ORTH : "gim"},  
                 {ORTH : "me"}                
    ])
```

```python
doc = nlp("gimme double cheese extra large healthy pizza" )
tokens = [token.text for token  in doc ]
tokens 
```

##### sentence tokenization

```python
doc = nlp( ''' "Dr. Smith arrived at 6:30 p.m. at the U.S. Embassy in London, U.K., to meet Mr. Jones. "Is the meeting still on?" he asked anxiously. Yes, it was! However, it was delayed by 15 mins. due to traffic. Visitiors can find updates at https://example.com. indices like 1.1, 1.2, etc., are listed on page 4. Miss. Gable said, "Wow! What a beautiful day..." before entering the building.
" ''')
```

```python
nlp.add_pipe("sentencizer")
```

```python
sentences =[]
for sentence in doc.sents:
    sentences.append(sentence)
sentences
```
