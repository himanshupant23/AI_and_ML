##### customization of blank pipeline 

```python
#Adding component from one pipeline to another 
source_nlp = spacy.load("en_core_web_sm")
nlp = spacy.blank("en")
nlp.add_pipe("ner" , source = source_nlp )
nlp.pipe_names # now its showing ner in blank pipeline 
```
