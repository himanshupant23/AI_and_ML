### numpy random 


```python
x= r.rand()
print(x)

y= r.rand(4)
print(y)

random_array = r.randint(100,size=(5))
print(random_array)

random_2D_array= r.randint(100 , size=(2,5))
print(random_2D_array)

random_value= r.choice(random_array)
print(random_value)

random_value_with_size= r.choice(random_array, size=(2,3))
print(random_value_with_size)


```
