###### subplots in matplotlib

```python
plt.subplot(2,2,1)
plt.subplot(2,2,4) # to plot in 4th box just plot after it 
plt.plot(x,y)
plt.subplot(2,2,3)
plt.plot(y,x)
plt.subplot(2,2,2)
plt.plot(x**2,x**2)
```

```python
plt.subplot(6,6,1)
plt.subplot(6,6,2)
plt.subplot(6,6,4)
plt.subplot(6,6,5)
plt.subplot(6,6,7)
plt.subplot(6,6,9)
```

```python
# object oriented way 
fig = plt.figure(figsize=(12,8))
axis1=fig.add_axes([0.1,0.1,1,1])
axis1.plot(x,y)
axis2 = fig.add_axes([0.3,0.2,.8,.8])
axis2.plot(x,y)
```

#### Distribution plots 


```python
import sys
!{sys.executable} -m pip install seaborn
```

```python
import seaborn as sns 
```

```python
tips=pd.read_csv("tips.csv")
tips
```

```python
# Qualitative data = sex,smoker,time,size,day
# quantative data = total_bill,tip

sns.displot(tips["total_bill"],bins=20)
```

```python
plt.subplot(1,2,1)
sns.histplot(tips["total_bill"])
plt.subplot(1,2,2)
sns.histplot(tips["tip"])
```

```python
# to compare both data we will use joint plot 
sns.jointplot(x = "total_bill", y = "tip", data = tips, kind = "scatter")
```

```python
sns.jointplot(x = "total_bill", y = "tip", data = tips, kind = "hex")
```

```python
# Pair plot  ----  it compare all the numerical data of dataframe(tips)
# with the help of "hue" we can read plot eg- we can categorize plots 
sns.pairplot(tips, hue="sex")
```

```python
# rugplots
sns.rugplot(tips["total_bill"])
```

##### Categorical plots

```python
# Count plots 
sns.countplot( x = tips["sex"], hue=tips["smoker"])
```

```python
# bar plot -- it takes two parameter x axis,y axis , y = quantative variable x= catgoricol variable 
sns.barplot( x = tips["sex"], y = tips["total_bill"])
# here mean of total_bill for male and female are shown in bar plot , we can also change  mean to another function like median ,sum , count etc . BY passing estimator parameter 
```

```python
sns.barplot( x = tips["sex"], y = tips["total_bill"], estimator = np.sum)
```

```python
sns.barplot( x = tips["sex"], y = tips["tip"], estimator = np.sum)
```

```python
# Box plot 
sns.boxplot(x = "tip", y="day",data = tips)
```

```python
# voilin plots
sns.violinplot(x = "day", y = "total_bill", data = tips)
```

```python
# stripplots and swarmplots

sns.stripplot(x ="day", y = "total_bill",data = tips)
```

```python
sns.swarmplot(x="day",y= "total_bill", data = tips)
```

```python
# combining swarmplots and violin plots
sns.violinplot(x= "tip",y="day",data=tips)
sns.swarmplot(x= "tip",y="day",data=tips)
```
